```python
# Python Functions - Theory and Practical Questions

## Theoretical Questions with Answers and Examples

### 1. What is the difference between a function and a method in Python?
"""Function: A block of reusable code that performs a specific task. It can be defined using def and used independently.

   Method: A function that is associated with an object (i.e., called on an object). Methods are defined within class definitions."""

# Function
def greet(name):
    return "Hello " + name

# Method
name = "Alice"
print(name.upper())  # upper() is a method


### 2. Explain the concept of function arguments and parameters in Python.

'''Parameters are variables listed in the function definition.
Arguments are the actual values passed to the function when it is called.'''

def add(a, b):  # a and b are parameters
    return a + b

print(add(3, 5))  # 3 and 5 are arguments

### 3. What are the different ways to define and call a function in Python?

'''You can define a function using def, and call it by its name followed by parentheses with arguments.'''

def square(x):
    return x * x

print(square(4))

square_lambda = lambda x: x * x
print(square_lambda(4))


### 4. What is the purpose of the `return` statement in a Python function?

'''The return statement sends back the result from the function to the caller and exits the function.'''

def multiply(a, b):
    return a * b

result = multiply(3, 4)
print(result)


### 5. What are iterators in Python and how do they differ from iterables?

'''Iterable: An object which can be iterated over (like list, tuple, set).

Iterator: An object with a __next__() method and __iter__() that keeps state and produces next value on demand.'''

lst = [1, 2, 3]  # iterable
it = iter(lst)  # iterator

print(next(it))
print(next(it))


### 6. Explain the concept of generators in Python and how they are defined.

'''Generators are a way to create iterators using functions and the yield statement.

They produce items one at a time and maintain state between iterations.'''

def count_up_to(n):
    i = 1
    while i <= n:
        yield i
        i += 1

for number in count_up_to(3):
    print(number)


### 7. What are the advantages of using generators over regular functions?

'''Memory Efficient: Generates items one by one using yield, doesn't store entire sequence in memory.

Lazy Evaluation: Items are generated only when needed.'''

def numbers():
    for i in range(1_000_000):
        yield i


### 8. What is a lambda function in Python and when is it typically used?

'''A lambda function is a small anonymous function defined using the lambda keyword.

Often used for short functions that are passed as arguments.'''

add = lambda x, y: x + y
print(add(2, 3))


### 9. Explain the purpose and usage of the `map()` function in Python.

'''map() applies a function to every item of an iterable and returns a map object (an iterator).'''

nums = [1, 2, 3, 4]
squares = map(lambda x: x * x, nums)
print(list(squares))


### 10. Difference between `map()`, `reduce()`, and `filter()`

'''map(func, iterable) – Applies func to all items.

filter(func, iterable) – Filters items where func(item) returns True.

reduce(func, iterable) – Repeatedly applies func to items cumulatively (from functools).'''

from functools import reduce
nums = [1, 2, 3, 4]
squares = list(map(lambda x: x * x, nums))
evens = list(filter(lambda x: x % 2 == 0, nums))
sum_total = reduce(lambda x, y: x + y, nums)
print(squares, evens, sum_total)


## Practical Questions with Code

### 1. Sum of all even numbers in a list

def sum_even_numbers(lst):
    return sum(num for num in lst if num % 2 == 0)

print(sum_even_numbers([1, 2, 3, 4, 5, 6]))


### 2. Reverse a string

def reverse_string(s):
    return s[::-1]

print(reverse_string("hello"))

### 3. Return list of squares

def square_list(lst):
    return [x ** 2 for x in lst]

print(square_list([1, 2, 3, 4]))


### 4. Prime check from 1 to 200

def is_prime(n):
    if n <= 1:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

primes = [i for i in range(1, 201) if is_prime(i)]
print(primes)


### 5. Fibonacci Iterator

class Fibonacci:
    def __init__(self, max_terms):
        self.max = max_terms
        self.a, self.b = 0, 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.max:
            raise StopIteration
        self.count += 1
        result = self.a
        self.a, self.b = self.b, self.a + self.b
        return result

for num in Fibonacci(10):
    print(num, end=' ')


### 6. Powers of 2 Generator

def powers_of_two(n):
    for i in range(n + 1):
        yield 2 ** i

for power in powers_of_two(5):
    print(power)


### 7. File Line Reader Generator

def read_file_line_by_line(filename):
    with open(filename, 'r') as file:
        for line in file:
            yield line.strip()

# Usage: for line in read_file_line_by_line("sample.txt"): print(line)


### 8. Sort tuples by second element

data = [(1, 3), (4, 1), (2, 2)]
sorted_data = sorted(data, key=lambda x: x[1])
print(sorted_data)


### 9. Celsius to Fahrenheit using map

celsius = [0, 20, 40, 100]
fahrenheit = list(map(lambda c: (c * 9/5) + 32, celsius))
print(fahrenheit)


### 10. Remove vowels using filter

def remove_vowels(s):
    return ''.join(filter(lambda x: x.lower() not in 'aeiou', s))

print(remove_vowels("Beautiful"))


### 11. Bookstore Accounting using map and lambda

orders = [
    ["34587", "Learning Python", 4, 40.95],
    ["98762", "Programming Python", 5, 56.80],
    ["77226", "Head First Python", 3, 32.95],
    ["88112", "Einführung in Python3", 3, 24.99]
]

result = list(map(
    lambda order: (order[0], order[2] * order[3] + (10 if order[2] * order[3] < 100 else 0)),
    orders
))

print(result)
```
